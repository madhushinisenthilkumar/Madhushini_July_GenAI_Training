```python
from google import genai
from google.colab import userdata

api_key = userdata.get("GEMINI_API_KEY")

client = genai.Client(api_key=api_key)

print("✅ Gemini connected successfully!")
```

```python
from typing import TypedDict


class Education(TypedDict):
    degree: str
    institute: str
    year: str
    score: str


class Experience(TypedDict):
    company: str
    role: str
    duration: str
    highlights: str


class Project(TypedDict):
    name: str
    description: str
    tech_stack: str


class Resume(TypedDict):
    name: str
    email: str | None
    phone: str | None
    linkedin_url: str | None
    education: list[Education]
    skills: list[str]
    experience: list[Experience]
    projects: list[Project]
    total_experience_years: float
    summary: str
```

```python
system_instruction = """
You are an expert Resume Parser.

Follow these rules carefully:

1. Extract only the information that is present in the resume.
2. Do NOT guess or invent any information.
3. If any field is missing, return null.
4. Normalize phone numbers to +91XXXXXXXXXX whenever possible.
5. Skills must be a clean list with no duplicates.
6. The summary must be exactly one sentence.
7. The resume may be written in English or any Indian language, but all output values must be in English.
8. Return only valid JSON.
"""
```

```python
def parse_resume(text: str):
    prompt = f"""
{system_instruction}

Resume:

{text}
"""

    response = client.models.generate_content(
        model="gemini-3.5-flash",
        contents=prompt
    )

    print(response.text)

    return response.text
```

```python
resume = """
ANJALI SHARMA

Email: anjali.sharma@gmail.com

Phone: 9876543210

Skills:
Python
Java
React
"""

parse_resume(resume)
```

```python
import json
from google.genai import types

def parse_resume(text: str):
    prompt = f"""
{system_instruction}

Resume:

{text}
"""

    response = client.models.generate_content(
        model="gemini-3.5-flash",
        contents=prompt,
        config=types.GenerateContentConfig(
            response_mime_type="application/json",
            response_schema=Resume
        )
    )

    try:
        return json.loads(response.text)

    except json.JSONDecodeError:
        print("❌ Invalid JSON returned.")
        print(response.text)
        return None
```

```python
def sanity_checks(data):
    print("===== Sanity Checks =====")

    # Check Name
    if data.get("name"):
        print("✅ Name Found")
    else:
        print("❌ Name Missing")

    # Check Email
    if data.get("email"):
        print("✅ Email Found")
    else:
        print("❌ Email Missing")

    # Check Skills
    if data.get("skills"):
        print(f"✅ {len(data['skills'])} Skills Found")
    else:
        print("❌ Skills Missing")

    # Check Experience
    if data.get("experience"):
        print("✅ Experience Found")
    else:
        print("⚠️ No Experience Mentioned")
```
